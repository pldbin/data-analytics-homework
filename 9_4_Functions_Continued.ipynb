2. Create a function `sum_numbers` that takes any number of numeric arguments and returns their sum. The function must work with any number of numeric arguments, including the case when there are no arguments.

  **Examples of function calls:**

  1. `sum_numbers(1, 2, 3, 4)` should return `10`.
  2. `sum_numbers(10, 20)` should return `30`.
  3. `sum_numbers()` should return `0`.

  **Implementation requirements:**

  - The function must use `*args` to accept any number of arguments.
  - The function must correctly handle the case when no arguments are passed.

  Demonstrate how the function works by running it on the sets below.

```python
def sum_numbers(*args):
    total = 0
    for number in args:
        total += number
    return total
```

```python
number_sets = [
    (1, 2, 3, 4),
    (10, 20),
    (11, 12, 13),
    (),
    (-19, -164, 183)
]
```

```python
for numbers in number_sets:
    print(numbers, "->", sum_numbers(*numbers))
```

3. **Refactoring task: calculating the Body Mass Index (BMI).** Refactoring is a standard practice when working with your own or someone else's code in a company. Here we'll learn what it is and practice doing it.

  You have a code fragment that calculates the [body mass index (BMI)](https://en.wikipedia.org/wiki/Body_mass_index) for five people. Your task is to refactor this code, i.e. improve how it looks by moving the repeated parts into a separate function. The goal of refactoring is to **keep the functionality of the existing code** while making it more readable and easier to understand and modify. To calculate BMI, divide the weight by the height squared.

  Move the duplicated code (code that performs the same operations, just with different numbers) into a function so that the result stays **exactly the same**, but there is less code.

  At the end, run your refactored code. If your code doesn't print the same output as the fragment before refactoring, the task is considered not completed.

  **Note**: this task is creative. You can wrap this code in a function in different ways. There is no single correct answer. What will be checked is your understanding of the approach of "moving repeated parts into a function".

```python
person1 = [80, 1.62]
person2 = [69, 1.53]
person3 = [80, 1.66]
person4 = [80, 1.79]
person5 = [72, 1.60]

bmi_person1 = int(person1[0] / person1[1]**2)
print("bmi {} = {}".format('person1', bmi_person1))

bmi_person2 = int(person2[0] / person2[1]**2)
print("bmi {} = {}".format('person2', bmi_person2))

bmi_person3 = int(person3[0] / person3[1]**2)
print("bmi {} = {}".format('person3', bmi_person3))

bmi_person4 = int(person4[0] / person4[1]**2)
print("bmi {} = {}".format('person4', bmi_person4))

bmi_person5 = int(person5[0] / person5[1]**2)
print("bmi {} = {}".format('person5', bmi_person5))
```

```python
def calculate_bmi(weight, height):
    return int(weight / height**2)


persons = {
    'person1': person1,
    'person2': person2,
    'person3': person3,
    'person4': person4,
    'person5': person5,
}

for name, person in persons.items():
    bmi = calculate_bmi(person[0], person[1])
    print("bmi {} = {}".format(name, bmi))
```
