1. **Counting up to a given number.**
   Task:
     - Create a `while` loop that counts from 1 to the number given in the variable `limit` (inclusive).
     - Print each number the counter goes through.
     - At the end, print the message "End of counting!".


```python
limit = 9  # Given number
```

```python
counter = 1
while counter <= limit:
    print(counter)
    counter += 1

print("End of counting!")
```

2. **Finding the first even number.** Task:
     - Create a `while` loop to find the first even number after the given number `start`.
     - Find and print the first even number.


```python
start = 7  # Given number
```

```python
number = start + 1
while number % 2 != 0:
    number += 1

print(number)
```


3. **Calculating a factorial.**   
  The factorial of a number is the product of all natural numbers from 1 up to that number inclusive. For example, the factorial of 5 (written as 5!) equals `1 * 2 * 3 * 4 * 5 = 120`.

  Task:  
      - Calculate the factorial of a given number using a `while` loop.  
      - Set a number for the factorial calculation in the variable `num`.  
      - Print the calculated factorial.  


```python
num = 5

factorial = 1
i = 1
while i <= num:
    factorial *= i
    i += 1

print(f"{num}! = {factorial}")
```
