5. The list `numbers` is defined below. Run the cell that defines it.

  Let's find the arithmetic mean of the elements of `numbers`, rounded to 2 decimal places, and store it in the variable `numbers_mean`. We'll do it in two simple steps.

  1) Find the sum of all elements of `numbers` and store it in the variable `numbers_sum`. You can use a loop or do it without one.

  2) Calculate `numbers_mean` by dividing the sum of the elements by their count. The number of elements in a list is its length.

  Print `numbers_mean` rounded to 2 decimal places.

  **Hint**: you can round a number with Python's built-in function [round()](https://docs.python.org/3/library/functions.html#round).

````python
numbers = [42, 51, 29, 2, 13, 44, 80, 66, 16]
````

````python
numbers_sum = 0
for number in numbers:
    numbers_sum += number

numbers_mean = round(numbers_sum / len(numbers), 2)
print(numbers_mean)
````

Expected result: `38.11`
