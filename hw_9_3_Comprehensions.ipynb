We'll use the following list for the next tasks:

````python
numbers = [42, 51, 29, 2, 13, 44, 80, 66, 16]
````

1. Using a `for` loop, create a new list `numbers_squared` where each element of `numbers` is multiplied by 2.

As a reminder: when we build a new list with a loop, we first create an empty list and then add a new element to it on each iteration.

In this case, the new element is the result of multiplying the current element of the original list `numbers` by 2.

````python
numbers_squared = []
for number in numbers:
    numbers_squared.append(number * 2)

print(numbers_squared)
````

Expected result: `[84, 102, 58, 4, 26, 88, 160, 132, 32]`

2. Using a `list comprehension`, create a new list `numbers_squared_lh` where each element of `numbers` is multiplied by 2.
Yes, the operation we perform on each element of the original list is the same as in the previous task, but the way we get the result is different.

````python
numbers_squared_lh = [number * 2 for number in numbers]
print(numbers_squared_lh)
````

3. We already have the mean of the list; it is stored in the variable `numbers_mean`.

  Using a loop, create a new list `numbers_gt_mean` that compares each element with this mean using the following rule: "if this element of `numbers` is greater than `numbers_mean`, write 1, otherwise 0". Print the new list.

````python
numbers_mean = 38.11
````

````python
numbers_gt_mean = []
for number in numbers:
    if number > numbers_mean:
        numbers_gt_mean.append(1)
    else:
        numbers_gt_mean.append(0)

print(numbers_gt_mean)
````

Expected result: `[1, 1, 0, 0, 0, 1, 1, 1, 0]`

4. Build a list using the same rule as in task 3, but now with a list comprehension. Store the result in the variable `numbers_gt_mean_lc`

````python
numbers_gt_mean_lc = [1 if number > numbers_mean else 0 for number in numbers]
print(numbers_gt_mean_lc)
````

5. A list `long` is given. Using a loop, create a dictionary `n_repeats` that shows "how many times each unique element of `long` occurs".
For example:
```
long = [1, 1, 2, 2, 3]
# then `n_repeats` contains {1: 2, 2: 2, 3: 1}
```

To find how many times each unique element occurs, use the list method `list.count(<element>)`.
Print `n_repeats`.

**Hints**:
1. It may be convenient to get the set of unique elements of the list first.
2. The order of keys in the dictionary doesn't matter! The keys don't have to be in the same order as in the example result. But the number of occurrences for each element must match.

````python
long = [75, 39, 95, 1, 91, 43, 97, 24, 86, 75, 92, 91, 15, 15, 75, 95, 54, 29, 55, 98]
````

````python
n_repeats = {}
for number in long:
    n_repeats[number] = long.count(number)

print(n_repeats)
````

Expected result:
```
{1: 1, 97: 1, 98: 1, 39: 1, 75: 3, 43: 1, 15: 2, 55: 1, 86: 1, 54: 1, 24: 1, 91: 2, 92: 1, 29: 1, 95: 2}

```

6. Do the same operation as in the previous task, but using a `dict comprehension`. Store the result in the variable `n_repeats_dh` and print it.

````python
n_repeats_dh = {number: long.count(number) for number in set(long)}
print(n_repeats_dh)
````
