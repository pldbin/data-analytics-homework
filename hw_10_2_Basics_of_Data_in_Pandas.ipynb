7.1. Using data filtering, find the mean value of `gross income` for the cities `Yangon` and `Mandalay`. In which city is it higher?

7.2. Formulate 3 hypotheses for why gross income is higher in one of the cities, how you could find out whether each hypothesis is true using data (existing and additional), and how the conclusions from these hypotheses could be used to make decisions on increasing profit in the supermarket chain. Write your answer as text in a separate Markdown cell.

This task is about product thinking. If you don't know which hypotheses to formulate, write the ones that seem completely primitive to you. It is important to learn to express your hypotheses about data, and with time and experience they will become deeper and more precise. All data analysis is built on testing certain hypotheses and drawing conclusions and recommendations based on your research.

```python
yangon_mean = df.loc[df["City"] == "Yangon", "gross income"].mean()
mandalay_mean = df.loc[df["City"] == "Mandalay", "gross income"].mean()

print(f"Yangon: {yangon_mean:.2f}")
print(f"Mandalay: {mandalay_mean:.2f}")
```

**7.1. Answer:** The mean gross income per invoice is higher in **Mandalay** (15.23 USD) than in **Yangon** (14.87 USD), by about 0.36 USD.

**7.2. Hypotheses**

**Hypothesis 1. Customers in Mandalay buy more expensive products.**
- *How to check:* compare the mean `Unit price` in the two cities, overall and within each `Product line`. Additional data: prices of specific products in each branch, to see whether the same products cost the same.
- *How to use it:* if Mandalay customers choose pricier items, widen the premium range there; in Yangon, test promoting higher-priced products (shelf placement, recommendations at checkout).

**Hypothesis 2. The product mix is different: Mandalay sells a bigger share of categories that bring more income per invoice.**
- *How to check:* compare the share of invoices by `Product line` in each city (task 9) and the mean `gross income` per category. Additional data: shelf space and stock levels by category in each store, and out-of-stock history.
- *How to use it:* give more shelf space and stock to the categories that earn the most in each city, and reduce deliveries of the ones that sell poorly.

**Hypothesis 3. Customers in Mandalay buy more items per visit (bigger baskets).**
- *How to check:* compare the mean `Quantity` per invoice in the two cities, and the mean `gross income` by `Customer type` (Member / Normal) within each city. Additional data: loyalty card history (how often members come back), store size, and the population and income levels of each city.
- *How to use it:* in Yangon, run multi-buy offers and bundles, and promote the membership card if members spend more.

**Before making decisions:** the difference is small (about 2.4%), so it is worth checking that it is stable from month to month (`Date`) and is not random. It also makes sense to compare total `gross income`, not only the mean, because Yangon has more invoices (task 5).
