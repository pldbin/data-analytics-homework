0. Import the pandas library and read the data into a variable `df` of type `pandas.DataFrame`.

```python
import pandas as pd

df = pd.read_csv(r"C:\HMPython\data_mat\supermarket_sales.csv")
```

1. Let's investigate which supermarket branch ('Branch') is the most profitable. To do this, find the total profit for each branch and identify which branch has the highest.

```python
branch_income = df.groupby("Branch")["gross income"].sum()

print(branch_income.round(2))
print("Most profitable branch:", branch_income.idxmax())
```

2. In which city is the branch with the highest profit located? Maybe we should open another store there.  
Find the answer using Pandas functionality.

```python
top_branch = branch_income.idxmax()
top_city = df.loc[df["Branch"] == top_branch, "City"].iloc[0]

print(f"Branch {top_branch} is in {top_city}")
```

3.1. Create a pivot table that shows how many purchases (invoices) were made in each branch (`Branch`) for the different product categories. Save the table to the variable `invoices_by_category` and print the variable.
This table will help analyze which branch buys the most products of each category.

```python
invoices_by_category = df.pivot_table(
    index="Branch",
    columns="Product line",
    values="Invoice ID",
    aggfunc="count",
)

invoices_by_category
```

**Question 1 (who and what):** Which product categories do women and men spend the most on?

```python
df.pivot_table(index="Product line", columns="Gender", values="Total", aggfunc="sum").round(2)
```

**Answer:** Women spend much more than men on Food and beverages (about 33,171 vs 22,974 USD), Fashion accessories (30,437 vs 23,868) and Home and lifestyle (30,037 vs 23,825). Men spend much more on Health and beauty (30,633 vs 18,561). In Electronic accessories and Sports and travel, the amounts are almost the same.

**How to use it:** target promotions and advertising by gender, for example offers on food and fashion accessories for women and on health and beauty products for men (such as men's care products). Personal offers for loyalty card holders can take this into account too.

**Question 2 (when):** On which day of the week do customers buy the most, and on which day the least?

```python
df["Weekday"] = pd.to_datetime(df["Date"], format="%m/%d/%Y").dt.day_name()

df.groupby("Weekday")["Total"].agg(["count", "sum"]).sort_values("sum", ascending=False).round(2)
```

**Answer:** Saturday is the strongest day: 164 purchases worth about 56,121 USD. Monday is the weakest: 125 purchases worth about 37,899 USD, which is about a third less than on Saturday.

**How to use it:** plan more staff at the checkouts and restock the shelves before Saturday (and Tuesday, the second-best day). To bring in more customers on the weakest days, run special offers on Mondays and Wednesdays.

**Question 3 (who and where):** Do members (customers with a membership card) spend more per purchase than normal customers, and is this the same in every city?

```python
df.pivot_table(index="City", columns="Customer type", values="Total", aggfunc="mean").round(2)
```

**Answer:** In Yangon and Mandalay, members spend more per purchase than normal customers (Yangon: about 321 vs 304 USD, roughly 6% more; Mandalay: about 326 vs 314 USD). In Naypyitaw there is almost no difference (about 337 vs 338 USD).

**How to use it:** in Yangon, which has the lowest average receipt, the chain can promote the membership card more actively, for example at checkout. In Naypyitaw, it is worth checking whether the card gives customers enough benefits. The differences are small, and members may simply be customers who already spend more, so a test (for example, offering the card to some of the customers) would show whether the card really increases spending.
