3.1. Create a new column `Rating Category` in the data that defines the rating category (`Rating`): high, medium, or low, according to the following rule:
- if the rating is greater than or equal to 8, it is high (`High`)
- if the rating is from 6 (inclusive) to 8 (exclusive), it is medium (`Medium`)
- otherwise, it is low (`Low`).

```python
def get_rating_category(rating):
    if rating >= 8:
        return "High"
    elif rating >= 6:
        return "Medium"
    else:
        return "Low"


df["Rating Category"] = df["Rating"].apply(get_rating_category)

df[["Rating", "Rating Category"]].head()
```

3.2. Print the percentage of the data that each rating category value makes up.

4.1. Run the code below to create the variable `time`. This variable contains a time in the "hour:minute" format. Write a function `get_hour` that takes a value in the same format as `time`, extracts only the hours, and converts them to the `int` type.

Run the function you wrote on the variable `time` and check that it works correctly.

```python
time = '11:28'
```

```python
def get_hour(time_value):
    hour = time_value.split(":")[0]
    return int(hour)


result = get_hour(time)

print(result, type(result))
```

4.2. Using the `get_hour` function and the `DataFrame.apply()` method, create a new column `Hour` in our dataset that contains the hour of purchase, calculated from the purchase time `Time`.

**Answer:** Customers make the most purchases at **19:00** (113 purchases), followed by **13:00** (103), **15:00** (102) and **10:00** (101). The fewest purchases are made at 17:00 (74) and 20:00 (75).

Why these hours, in my view:

- **10:00** – the store opens at 10:00, so people who do not work during the day can come right away.

Is it logical? Mostly yes: the evening peak after work and the lunch peak match a usual daily routine. To be sure about the pattern, it would be worth checking whether it repeats on different days of the week and in each city.
