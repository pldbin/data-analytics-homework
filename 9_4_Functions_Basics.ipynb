5. **Division with type checking.**  

  1. Write a function `int_division` that takes two `int` arguments, `a` and `b`, and divides the first argument by the second: `a/b`. Before dividing, the function checks whether the second argument is not equal to 0. If the second argument isn't 0, the function returns the result of the division; otherwise, instead of dividing, it displays the message "Cannot divide by 0".

    If the arguments passed to the function are not of type `int`, the function prints the message "I don't work with these data types.".

  2. Run the function on the following pairs of numbers:
- 12 and 6
- 20 and 0
- 112 and 10
- 99 and 0
- 11 and 1.1

    and print the message "Function result: " followed by what the function returned.

  3. Answer the question: "Why did the function return None in some cases?"

````python
def int_division(a, b):
    if type(b) != int:
        print("I don't work with these data types.")
    elif b == 0:
        print("Cannot divide by 0")
    else:
        return a / b


pairs = [(12, 6), (20, 0), (112, 10), (99, 0), (11, 1.1)]
for a, b in pairs:
    result = int_division(a, b)
    print("Function result:", result)
````

**Answer to question 3:** in some cases the function only prints a message (division by 0 or wrong data types) and never reaches `return`. If a function ends without `return`, Python returns `None` by default, so we see `None` as the result.
