# Homework: "if...else Statement"

1. **Temperature check:**
   - Task:
     - A predefined variable `temperature` is given. Check its value.
     - If the temperature is above 25 degrees Celsius, print "It's a warm day."
     - If the temperature is between 15 and 25 degrees Celsius (inclusive), print "It's a pleasant day."
     - Otherwise, print "It's a cold day."

```python
temperature = 20
```

```python
if temperature > 25:
    print("It's a warm day.")
elif 15 <= temperature <= 25:
    print("It's a pleasant day.")
else:
    print("It's a cold day.")
```

2. **Grade calculator:**
   - Task:
     - A predefined variable `score` holds an exam result (out of 100 points). Your task is to determine the corresponding letter grade using the rules below.
     - If the score is 90 or higher, print "A".
     - If the score is between 80 and 89, print "B".
     - If the score is between 70 and 79, print "C".
     - If the score is between 60 and 69, print "D".
     - Otherwise, print "F".
     


```python
score = 75
```

```python
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```



3. **Leap year check:**
   - Task:
     - A predefined variable `year` is given. Check whether it is a leap year. The rules are listed below.
     - If the year is divisible by 4 and not divisible by 100, or it is divisible by 400, print "Leap year".
     - Otherwise, print "Not a leap year".


```python
year = 2024  # Example year
```

```python
if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```
